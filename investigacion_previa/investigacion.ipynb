```python
import pandas as pd 

bank_raw = pd.read_csv('../transacciones.csv',  sep=";")

bank_column_types = bank_raw.dtypes

print(f'Tipo de dato de los campos: {bank_column_types}\n')

print(f"""Valores únicos por campo:\n id_transaccion: {bank_raw['id_transaccion'].unique()}
\n fecha_hora: {bank_raw['fecha_hora'].unique()}
\n id_cuenta: {bank_raw['id_cuenta'].unique()}
\n tipo: {bank_raw['tipo'].unique()}
\n monto: {bank_raw['monto'].unique()}
\n moneda: {bank_raw['moneda'].unique()}
\n canal {bank_raw['canal'].unique()}
\n id_sucursal: {bank_raw['id_sucursal'].unique()}
\n estado: {bank_raw['estado'].unique()}\n""")

print(f"""Cantidad de valores duplicados en el siguiente campo:\n id_transaccion: {bank_raw['id_transaccion'].duplicated().sum()}\n""")

print(f"""Valores nulos por campo:\n id_transaccion: {bank_raw['id_transaccion'].isnull().sum()}
\n fecha_hora: {bank_raw['fecha_hora'].isnull().sum()}
\n id_cuenta: {bank_raw['id_cuenta'].isnull().sum()}
\n tipo: {bank_raw['tipo'].isnull().sum()}
\n monto: {bank_raw['monto'].isnull().sum()}
\n moneda: {bank_raw['moneda'].isnull().sum()}
\n canal {bank_raw['canal'].isnull().sum()}
\n id_sucursal: {bank_raw['id_sucursal'].isnull().sum()}
\n estado: {bank_raw['estado'].isnull().sum()}\n""")

date_formats = (
    bank_raw['fecha_hora']
    .str.replace(r"\d", "9", regex=True)
    .value_counts()
)

print(f'Cantidad de valores en campo fecha_hora por formato: \n{date_formats}')
```
